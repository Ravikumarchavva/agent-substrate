```python
import subprocess

subprocess.run([
    "mineru",
    "-p", "image.png",
    "-o", "output",
    "-b", "pipeline"
])
```

```python
from paddleocr import PPStructureV3

pipeline = PPStructureV3(lang='en', device='gpu')

result = pipeline.predict("small.png")

print(result)
```

```python
for page in result:
    page.save_to_markdown(save_path='ppoutput')
```
